# Data Structure

A data structure is a way of **organizing** and **storing data** so that it can be accessed and modified efficiently. Think of it as a container that holds data in a specific format, optimized for certain operations like searching, inserting, deleting, or sorting.


## Purpose
Makes data easier to use, process, and manage.

## Types
There are two main categories:

### 1. Primitive
Basic types like:
- Integers
- Floats
- Characters

### 2. Non-Primitive / Abstract Data Structures
Complex structures like:
- Arrays
- Linked Lists
- Stacks
- Queues
- Trees
- Graphs
- Hash Tables

## Operations
Depending on the structure, you can perform operations such as:
- Insert
- Delete
- Search
- Traverse
- Sort

## Data Structure Examples

### Array
- Stores elements in contiguous memory locations.
- Easy to access by index.
- Costly to insert or delete elements in the middle.

### Linked List
- Stores elements as nodes linked via pointers.
- Easy to insert and delete elements anywhere.
- Slower to access specific elements compared to arrays.

# 1️⃣ Linear Data Structures


These store elements sequentially.


## 1. Array (List in Python)
- An array is a data structure that stores multiple elements of the same type in contiguous memory locations.
- Each element is accessed using an index (position).

**Key Properties**
1. Indexed access → O(1)
2. Fixed size (in low-level languages)
3. Python list is dynamic
4. Contiguous Memory
 

##  Array: Time Complexity & Use Cases

### Time Complexity

| Operation        | Complexity       |
|-----------------|----------------|
| Access           | O(1)           |
| Insert (end)     | O(1) amortized |
| Insert (middle)  | O(n)           |
| Delete           | O(n)           |

### Use Cases
- Storing ordered data
- Lookup by index
- Base structure for many algorithms

# Types of Arrays

## 1. One-Dimensional Array
- Stores elements in a single linear sequence.
- Accessed using a single index.
- Example: `arr[5]`

## 2. Two-Dimensional Array
- Stores data in a table-like structure (rows and columns).
- Accessed using two indices: `arr[row][col]`.
- Example: `matrix[3][4]`

## 3. Multi-Dimensional Array
- Extends beyond 2 dimensions (3D, 4D, …).
- Used in complex data representations like matrices, tensors.
- Example: `tensor[2][3][4]`

## 4. Dynamic Array
- Size can grow or shrink during runtime.
- Example: Python `list`, Java `ArrayList`.

## 5. Jagged Array
- Each row can have different number of columns.
- Example: `arr = [[1,2,3], [4,5], [6]]` (Python)

# Linked List


A linked list is a linear data structure where elements, called nodes, are not stored in contiguous memory (like arrays) but are connected via pointers. Each node contains:

1. Data – the value it holds.

2. Next (or Pointer/Reference) – a reference to the next node in the sequence

| Type                 | Description                                              |
|----------------------|----------------------------------------------------------|
| **Singly Linked List**    | Each node points to the next node. Traversal is one-way. |
| **Doubly Linked List**    | Each node has pointers to both the next and previous nodes. Traversal both ways. |
| **Circular Linked List**  | The last node points back to the first node, forming a loop. |

## Key Points

- **Dynamic size** – you can easily add or remove nodes without shifting elements (unlike arrays).  
- **Insertion/Deletion** – efficient if you have a pointer to the node.  
- **Access** – slower than arrays for random access; you must traverse nodes sequentially.  
- **Non-contiguous Memory**  –  Nodes are stored anywhere in memory and connected using pointers.


### Visual Example (Singly Linked List)

> [Data|Next] -> [Data|Next] -> [Data|Next] -> None


### Visual Example (Doubly Linked List (DLL))

> None <- [10|Next|Prev] <-> [20|Next|Prev] <-> [30|Next|Prev] -> None

### Visual Example (Circular Linked List (CLL))

>[10|Next|Prev] <-> [20|Next|Prev] <-> [30|Next|Prev]
>
>       ^------------------------------------------+

## Basic Operations (Linked List)

| Operation | Description | Time Complexity |
|-----------|-------------|----------------|
| Access | Traverse to element | O(n) |
| Insert (beginning) | Change head pointer | O(1) |
| Insert (middle) | Adjust links (if node known) | O(1) |
| Delete | Adjust links (if node known) | O(1) |
| Search | Traverse nodes | O(n) |

#### Where Linked Lists Are Used

Common systems using linked lists:

- Implement Stacks

- Implement Queues

- Hash table chaining

- Graph adjacency list

- Memory management

- Browser history

- Music playlist navigation

## Stack
- A stack is a linear data structure that follows the LIFO principle (Last In, First Out).
- This means the last element added is the first one to be removed.

- Think of it like a stack of plates: you put a plate on top and also remove from the top

## Basic Stack Operations

| Operation   | Meaning |
|-------------|---------|
| Push        | Insert an element on the top of the stack |
| Pop         | Remove the top element |
| Peek / Top  | View the top element without removing |
| isEmpty     | Check if the stack has no elements |
| isFull      | Check if stack is full (mainly in fixed-size implementations) |.

####  Implementation Methods
1. Array

2. Linked List

##### 1. Stack Using Array

- Elements stored in contiguous memory

- Access using index

- Needs fixed size in some languages.

##### 2. Stack Using Linked List

- Each node contains data + pointer

- Dynamic size

- No fixed capacity

#### Time Complexity of Stack Operations

| Operation | Time Complexity |
|-----------|-----------------|
| Push      | O(1)            |
| Pop       | O(1)            |
| Peek      | O(1)            |

### Common Interview Use Cases

Stacks are heavily used in algorithms such as:

- Balanced Parentheses

- Undo / Redo operations

- Expression evaluation

- DFS (Depth First Search)

- Backtracking algorithms
